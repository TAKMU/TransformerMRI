```python
import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor
import pandas as pd
from tqdm import tqdm
from convert import convert_patient

df = pd.read_csv("./data/labels_duke.csv")

with ProcessPoolExecutor(max_workers=16, mp_context=mp.get_context("fork")) as ex:
    results = list(tqdm(ex.map(convert_patient, df["image"]), total=len(df)))

df["patient"] = [r[0] for r in results]
df["nifti"] = [r[1] for r in results]
errors = [(r[0], r[2]) for r in results if r[2]]
print(f"{len(errors)} failures", errors[:5])

df.to_csv("./data/labels_duke_nifti.csv", index=False)
```
